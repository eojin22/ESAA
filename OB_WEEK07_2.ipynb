# 5장. 합성곱 신경망

## **5-1. 합성곱 신경망**
- 딥러닝에서는 순전파 과정에 따라 계산된 오차 정보가 역전파를 통해 신경망의 모든 노드(출력층 → 은닉층 → 입력층)로 전송됨. 이러한 계산 과정은 복잡하고 많은 자원(CPU, GPU, 메모리)을 요구하며 계산 시간도 오래 걸림
- 이러한 문제를 해결하고자 등장한 것이 합성곱 신경망
- 합성곱 신경망(Convolutional Neural Network, CNN/ConvNet): 이미지 전체를 한 번에 계산하는 것이 아니라 이미지의 국소적 부분을 계산함으로써 시간과 자원을 절약하고, 이미지의 세밀한 부분까지 분석할 수 있는 신경망

### **5-1-1. 합성곱층의 필요성**
- 합성곱 신경망은 이미지나 영상을 처리하는 데 유용함
- 일반적인 신경망에서 이미지를 분석할 때는 이미지의 픽셀을 1차원 배열로 펼쳐(Flattening) 각 픽셀에 가중치를 곱한 후 은닉층으로 전달함
- 하지만 이미지를 펼쳐서 분석하면 데이터의 공간적 구조를 무시하게 되는데, 이러한 문제를 방지하기 위해 합성곱층이 도입됨

### **5-1-2. 합성곱 신경망 구조**
- 합성곱 신경망은 음성 인식이나 이미지·영상 인식에서 주로 사용되는 신경망
- 다차원 배열 데이터를 처리하도록 구성되어 컬러 이미지와 같은 다차원 배열 처리에 특화되어 있음
- 입력층 / 합성곱층 / 풀링층 / 완전연결층 / 출력층의 다섯 개 계층으로 구성됨
- 합성곱 신경망은 합성곱층과 풀링층을 거치면서 입력 이미지의 주요 특성 벡터(Feature Vector)를 추출함. 이후 추출된 주요 특성 벡터들은 완전연결층을 거치면서 1차원 벡터로 변환되며, 마지막으로 출력층에서 활성화 함수인 소프트맥스(Softmax) 함수를 사용하여 최종 결과가 출력됨

- **입력층(Input Layer):** 입력 이미지 데이터가 최초로 거치는 계층. 이미지는 단순한 1차원 데이터가 아니라 높이(Height), 너비(Width), 채널(Channel)의 값을 갖는 3차원 데이터임. 이때 채널은 이미지가 그레이스케일이면 1개, 컬러 이미지이면 RGB의 3개를 가짐. 예를 들어 높이 4, 너비 4인 RGB 이미지는 (4, 4, 3)으로 표현됨

- **합성곱층(Convolutional Layer):** 입력 데이터에서 특성을 추출하는 역할. 입력 이미지의 특성을 감지하기 위해 커널(Kernel) 또는 필터(Filter)를 사용하며, 커널/필터는 이미지의 모든 영역을 훑으면서 특성을 추출함. 이렇게 추출된 결과물이 특성 맵(Feature Map)임. 커널은 일반적으로 3×3, 5×5 크기로 적용되며 스트라이드(Stride)라는 지정된 간격에 따라 순차적으로 이동함

- 합성곱 연산은 입력 이미지와 필터를 겹쳐 놓고 대응되는 숫자끼리 곱한 후 모두 더하는 방식으로 진행됨
- 필터는 스트라이드 간격만큼 이동하면서 합성곱 연산을 반복하고, 새로운 특성 맵을 생성함
 - 예를 들어 (6, 6, 1) 크기의 이미지에 3×3 필터를 스트라이드 1로 적용하면 (4, 4, 1) 크기의 특성 맵이 생성됨
 - 컬러 이미지의 경우 RGB 각각에 서로 다른 가중치로 합성곱 연산을 적용한 후 결과를 더함. 이때 필터의 채널이 3개라고 해서 필터 개수가 3개인 것은 아니며, 하나의 필터가 3개 채널에 걸쳐 적용되는 것임




- **풀링층(Pooling Layer):** 합성곱층과 유사하게 특성 맵의 차원을 다운샘플링(Downsampling)하여 연산량을 감소시키고, 주요한 특성 벡터를 추출하여 학습을 효과적으로 할 수 있게 하는 계층. 다운샘플링은 이미지의 크기를 축소하는 것을 의미함. 풀링 연산에는 최대 풀링과 평균 풀링 두 가지가 사용됨
- **최대 풀링(Max Pooling)** : 대상 영역에서 최댓값을 추출하는 방법. 예를 들어 3, -1, -3, 1 중에서 최댓값인 3을 선택함
- **평균 풀링(Average Pooling)** : 대상 영역의 값들을 평균 내어 반환하는 방법. 예를 들어 3, -1, -3, 1의 평균인 0을 반환함
- 대부분의 합성곱 신경망에서는 최대 풀링이 사용되는데, 평균 풀링은 각 커널 값을 평균화하여 중요한 가중치를 갖는 값의 특성이 희미해질 수 있기 때문임
- 최대 풀링과 평균 풀링은 계산 방법은 다르지만 사용하는 하이퍼파라미터는 동일함

- **완전연결층(Fully Connected Layer)**: 합성곱층과 풀링층을 거치면서 차원이 축소된 특성 맵이 최종적으로 전달되는 계층. 이 과정에서 이미지는 3차원 벡터에서 1차원 벡터로 펼쳐지게(Flatten) 됨

- **출력층(Output Layer)**: 소프트맥스(Softmax) 활성화 함수가 사용되며, 입력받은 값을 0~1 사이의 값으로 출력함. 마지막 출력층의 소프트맥스 함수를 통해 이미지가 각 레이블(Label)에 속할 확률 값이 출력되며, 가장 높은 확률 값을 갖는 레이블이 최종 결과로 선정됨

### **5-1-3. 1D, 2D, 3D 합성곱**
- 합성곱은 필터가 이동하는 방향의 수와 출력 형태에 따라 1D, 2D, 3D로 분류됨

- **1D 합성곱(1D Convolution)**: 필터가 시간 축을 따라 좌우로만 이동하는 합성곱. 입력(W)과 필터(k)에 대한 출력은 1D 배열 형태가 됨. 예를 들어 입력이 [1, 1, 1, 1, 1]이고 필터가 [0.25, 0.5, 0.25]라면 출력은 [1, 1, 1]이 됨. 그래프 곡선을 완화할 때 많이 사용됨

- **2D 합성곱(2D Convolution)**: 필터가 가로와 세로의 두 방향으로 움직이는 합성곱. 입력(W, H)에 필터(k, k)를 적용하면 출력은 (W, H)의 2D 행렬 형태가 됨

- **3D 합성곱(3D Convolution)**: 필터가 가로, 세로, 깊이의 세 방향으로 움직이는 합성곱. 입력(W, H, L)에 필터(k, k, d)를 적용하면 출력은 (W, H, L)의 3D 형태가 됨. 교재에서는 이때 필터 깊이 d가 입력 깊이 L보다 작은 조건(d < L)을 설명함

- **3D 입력을 갖는 2D 합성곱**: 입력이 (224×224×3, 112×112×32)와 같은 3D 형태임에도 출력은 필터 하나를 기준으로 2D 행렬 형태를 갖는 합성곱. 필터의 깊이는 입력 채널의 깊이와 같아야 하므로 입력(W, H, L)에 필터(k, k, L)를 적용함. 필터는 가로와 세로의 두 방향으로 이동하며 출력은 (W, H)의 2D 행렬 형태가 됨. 대표적인 사례로는 LeNet-5와 VGG가 있음

- **1×1 합성곱(1×1 Convolution)**: 3D 형태의 입력(W, H, L)에 필터(1, 1, L)를 적용하는 합성곱. 출력은 필터 하나를 기준으로 (W, H)의 2D 형태가 됨. 1×1 합성곱은 채널 수를 조정하여 연산량을 감소시키는 효과가 있으며, 대표적인 사례로는 GoogLeNet이 있음

# **0907 수상작 리뷰**
## 주제 및 데이터

주제: 문맥 기반 문장 순서 예측 AI 경진대회

**데이터 구성:**

- 여러 개의 문장이 주어졌을 때, 문맥을 고려하여 가장 자연스러운 문장 순서를 예측하는 문제
- 주어진 문장들의 의미적·언어적 연결 관계를 파악하여 올바른 순서를 결정
- 언어 모델이 특정 문장 순서를 얼마나 자연스럽게 판단하는지를 Perplexity(PPL)를 이용하여 비교
- 수상작에서는 별도의 분류 모델만 사용하는 것이 아니라 Causal Language Model의 특성을 활용하여 문장 순서를 예측함

## 코드 리뷰
1. Causal Language Model을 활용한 문장 순서 예측
- 문장 순서 예측을 단순한 분류 문제로 접근하지 않고, Causal Language Model이 다음 토큰을 예측하는 과정을 활용
- 여러 문장 순서를 모델에 입력한 후 각각의 Loss와 PPL을 계산
- 문맥상 자연스러운 순서일수록 언어 모델이 더 잘 예측할 수 있기 때문에 PPL이 낮게 나타날 것이라는 점을 이용하여 최종 순서를 결정

2. PPL(Perplexity)을 활용한 후보 순서 비교
- 주어진 문장들의 가능한 순서를 후보로 생성
- 각 후보 문장 순서를 하나의 텍스트로 연결하여 언어 모델에 입력
- 모델이 계산한 Loss를 기반으로 PPL을 구하고, 후보들의 PPL을 비교
- 가장 낮은 PPL을 갖는 문장 순서를 가장 자연스러운 순서로 선택

3. 다양한 LLM 및 Fine-tuning 실험
- 하나의 모델에만 의존하지 않고 여러 LLM을 활용하여 성능을 비교
- 상대적으로 작은 모델에서는 Full Parameter Tuning을 적용
- 대규모 모델에서는 QDoRA를 활용하여 파라미터 효율적인 Fine-tuning을 수행
- 다양한 모델과 학습 방법을 실험하면서 문제에 적합한 조합을 탐색

4. 여러 모델을 활용한 앙상블
- 단일 모델의 예측 결과만 사용하는 것에서 그치지 않고 여러 모델의 결과를 결합
- 서로 다른 모델이 가진 예측 특성을 활용하여 최종 성능을 높임
- 이러한 과정을 통해 Public과 Private 모두에서 높은 성능을 달성

### 코드 흐름 요약
문장 데이터 로드 → 문장 순서 후보 생성 → 토크나이저를 이용한 토큰화 → 인과적 언어 모델(Causal Language Model)에 입력 → 각 후보의 손실값 계산 → PPL(퍼플렉서티) 계산 → 후보별 PPL 비교 → 가장 낮은 PPL을 갖는 문장 순서 선택 → 파인튜닝을 통한 성능 개선 → 여러 모델의 결과 앙상블 → 최종 문장 순서 예측

### 주요 코드

```python
# TF-IDF 벡터화 코드
tfidf_vect = TfidfVectorizer(tokenizer=LemNormalize,
                             stop_words='english',
                             ngram_range=(1, 2),
                             min_df=0.05,
                             max_df=0.85)

feature_vect = tfidf_vect.fit_transform(document_df['opinion_text'])
```

## 차별점 및 배울 점
**1. PPL(Perplexity)의 의미와 활용 방법**
- PPL은 여러 문장의 순서를 비교하는 기준으로도 사용할 수 있다는 점을 알게 됨.
- 특히 PPL이 낮을수록 모델이 해당 문맥을 더 자연스럽게 예측한다는 원리를 실제 문제 해결에 적용한 것이 흥미로웠음.

**2. Fine-tuning 방법에도 여러 가지 접근 방식이 있다는 것을 알게 됨.**
- 작은 모델에는 Full Parameter Tuning을 적용하고, 대규모 모델에는 QDoRA와 같은 방법을 사용할 수 있다는 점을 배움.
- 모델의 크기와 컴퓨팅 자원에 따라 적절한 학습 방법을 선택하는 것도 중요한 부분이라고 느낌.
